```python
import pandas as pd
import os
from pathlib import Path

from scripts.build_mapping import build_mapping, canonical
```

```python
def add_cell_type(df: pd.DataFrame, mapping: dict[str, str]) -> pd.DataFrame:
    df["cell.type"] = df["phenotype"].map(lambda p: mapping.get(canonical(p), "unknown"))

    return df
```

```python
mapping = build_mapping(Path(f"tsv/IF1_phen_to_cell_mapping.csv"))
```

```python
directory = f"tsv/cells_properties/"

for filename in os.listdir(directory):
    print(f"[{filename}] starting...")

    df_tmp_raw = pd.read_csv(os.path.join(directory, filename), sep="\t", compression="gzip")

    df_tmp = add_cell_type(df_tmp_raw, mapping)

    df_tmp.to_csv(f"files_with_cell_types/{filename}")
```
